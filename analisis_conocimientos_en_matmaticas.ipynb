```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

gap_ds = pd.read_csv(
    'https://raw.githubusercontent.com/holtzy/The-Python-Graph-Gallery/master/static/data/gapminderData.csv'
)
```

```python
gap_ds
```

```python
datos_filtrados = gap_ds.loc[gap_ds.year == 2007]
```

```python
print(gap_ds.shape)
print(datos_filtrados.shape)
```

```python
sns.set_theme(style="darkgrid", palette="pastel")

sns.scatterplot(
    data=datos_filtrados,
    x='gdpPercap',
    y='lifeExp',
    sizes= (20,2000),
    size='pop',
    )

plt.title('Graficos simples de tipo Buelble Chart')
plt.show()
```

```python
sns.set_theme(style="darkgrid")
sns.scatterplot(
    data=datos_filtrados,
    x='gdpPercap',
    y='lifeExp',
    hue='continent',
    size='pop',
    alpha=0.5,
    sizes=(20, 400),
)

plt.title('grafico de tipo Buble Chart cin manejo de colores')
plt.show()
```

```python
Estudiantes_ds = pd.read_csv('E4_StudentsPerformance.csv')
```

```python
Estudiantes_ds
```

```python
sns.set_palette("Paired")
ax = sns.boxplot(x="gender", y="math score", data=Estudiantes_ds)
ax.set_xlabel("Gender")
ax.set_ylabel("Math Score")
plt.title('Gráficos de cajas de score matematico pr genero')
plt.show()
```

```python
sns.set_palette("pastel")
ax = sns.boxplot(x="gender", y="math score", hue = 'lunch', data=Estudiantes_ds)
ax.set_xlabel("Gender")
ax.set_ylabel("Math Score")
plt.title('Gráficos de cajas de score matematico pr genero/ lunch')
plt.show()
```

```python
sns.barplot(Estudiantes_ds, x ="race/ethnicity", y="math score")
```

```python
puntaje = Estudiantes_ds.groupby(['race/ethnicity', 'gender']) ['math score'].sum().unstack().fillna(0)
puntaje
```

```python
#Graficar una stacked var con pandas
puntaje.plot(kind='bar', stacked=True)

#Agregamos titulo definicion de niveles de rotacion
plt.xticks(rotation=0, ha='center')
```
